```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
import joblib
from pathlib import Path
import gc
```

```python
data = np.load("../data/relativistic_dataset_1m.npy")

X = data[:, [0, 1, 2, 3, 4]]
Y = data[:, 5:13]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
)
```

```python
names = [
    "position",
    "velocity",
    "beta",
    "gamma",
    "momentum",
    "kinetic_energy",
    "total_energy",
    "proper_time",
]
```

```python
model = DecisionTreeRegressor(
    max_depth=20,
    random_state=42,
)

model.fit(X_train, Y_train)
joblib.dump(model, "../models/decision_tree.joblib")
print("Saved: ../models/decision_tree.joblib")
Y_pred = model.predict(X_test)
```

```python
print("DECISION TREE REGRESSION")
for i, name in enumerate(names):

    rmse = np.sqrt(
        mean_squared_error(
            Y_test[:, i],
            Y_pred[:, i],
        )
    )

    r2 = r2_score(
        Y_test[:, i],
        Y_pred[:, i],
    )

    print(
        f"{name:16s} "
        f"RMSE = {rmse:.6e}   "
        f"R² = {r2:.6f}"
    )   
```

```python
del model
gc.collect()
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=20,
    random_state=42,
    n_jobs=-1,
)

model.fit(X_train, Y_train)
joblib.dump(model, "../models/RandomForestRegressor.joblib")
print("Saved: models/RandomForestRegressor.joblib")
```

```python
Y_pred = model.predict(X_test)
print("RANDOM FOREST REGRESSION")
for i, name in enumerate(names):

    rmse = np.sqrt(
        mean_squared_error(
            Y_test[:, i],
            Y_pred[:, i],
        )
    )

    r2 = r2_score(
        Y_test[:, i],
        Y_pred[:, i],
    )

    print(
        f"{name:16s} "
        f"RMSE = {rmse:.6e}   "
        f"R² = {r2:.6f}"
    )
```

```python
del model
gc.collect()
```

```python
model = ExtraTreesRegressor(
    n_estimators=100,
    max_depth=20,
    random_state=42,
    n_jobs=-1,
)
model.fit(X_train, Y_train)
joblib.dump(model, "../models/ExtraTreesRegressor.joblib")
print("Saved: models/ExtraTreesRegressor.joblib")
```

```python
Y_pred = model.predict(X_test)
print("EXTRA TREES REGRESSION")
for i, name in enumerate(names):
    rmse = np.sqrt(
        mean_squared_error(Y_test[:, i], Y_pred[:, i])
    )

    r2 = r2_score(
        Y_test[:, i],
        Y_pred[:, i]
    )

    print(
        f"{name:16s} "
        f"RMSE = {rmse:.6e}   "
        f"R² = {r2:.6f}"
    )
```

```python
del model
gc.collect()
```

```python
print("GRADIENT BOOSTING REGRESSION")
grad_models = []
for i, name in enumerate(names):

    model = GradientBoostingRegressor(
        n_estimators=100,
        max_depth=5,
        learning_rate=0.1,
        random_state=42,
    )

    model.fit(
        X_train,
        Y_train[:, i],
    )

    grad_models.append(model)

    Y_pred[:, i] = model.predict(X_test)

    rmse = np.sqrt(
        mean_squared_error(
            Y_test[:, i],
            Y_pred[:, i],
        )
    )

    r2 = r2_score(
        Y_test[:, i],
        Y_pred[:, i],
    )

    print(
        f"{name:16s} "
        f"RMSE = {rmse:.6e}   "
        f"R² = {r2:.6f}"
    )
```

```python
joblib.dump(
    grad_models,
    "../models/gradient_boosting.joblib"
)
del model
del grad_models
gc.collect()
```

```python
print("HISTOGRAM GRADIENT BOOSTING REGRESSION")
hist_models = []
for i, name in enumerate(names):

    model = HistGradientBoostingRegressor(
        max_iter=100,
        max_leaf_nodes=31,
        learning_rate=0.1,
        random_state=42,
    )

    model.fit(
        X_train,
        Y_train[:, i],
    )
    hist_models.append(model)

    Y_pred[:, i] = model.predict(X_test)

    rmse = np.sqrt(
        mean_squared_error(
            Y_test[:, i],
            Y_pred[:, i],
        )
    )

    r2 = r2_score(
        Y_test[:, i],
        Y_pred[:, i],
    )

    print(
        f"{name:16s} "
        f"RMSE = {rmse:.6e}   "
        f"R² = {r2:.6f}"
    )
joblib.dump(
    hist_models,
    "../models/hist_gradient_boosting.joblib"
)
print("\nSaved: hist_gradient_boosting.joblib")    
```

```python
del model
del hist_models
gc.collect()
```
