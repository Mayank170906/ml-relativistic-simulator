```python
import numpy as np
import pandas as pd
```

```python
train=np.load('../data/relativistic_dataset_1m.npy')
train.shape
```

```python
test=np.load('../data/relativistic_extreme_test_10k.npy')
test.shape
```

```python
columns = [
    "mass", "force", "initial_velocity", "initial_beta", "time",
    "position", "velocity", "beta", "gamma", "momentum",
    "kinetic_energy", "total_energy", "proper_time"
]
```

```python
train_df = pd.DataFrame(train, columns=columns)
test_df = pd.DataFrame(test, columns=columns)
```

```python
print("Train Summary:")
train_df.describe(include='all')
```

```python
print("\nTest Summary:")
test_df.describe(include='all')
```

```python
train_median = train_df.median()
test_median = test_df.median()
```

```python
train_mode = train_df.mode().iloc[0]
test_mode = test_df.mode().iloc[0]
```

```python
train_quartiles = train_df.quantile([0.25, 0.5, 0.75])
test_quartiles = test_df.quantile([0.25, 0.5, 0.75])
```

```python
print("\nTrain Median:\n", train_median)
print("\nTest Median:\n", test_median)
print("\nTrain Mode:\n", train_mode)
print("\nTest Mode:\n", test_mode)
print("\nTrain Quartiles:\n", train_quartiles)
print("\nTest Quartiles:\n", test_quartiles)
```

```python
train_df['gamma'].plot.hist(bins=50, alpha=0.5, label='Train', title="Gamma Distribution")
test_df['gamma'].plot.hist(bins=50, alpha=0.5, label='Test')
```

```python
train_df['momentum'].plot.hist(bins=50, alpha=0.5, label='Train', title="Momentum Distribution")
test_df['momentum'].plot.hist(bins=50, alpha=0.5, label='Test')
```

```python
train_df[['gamma','momentum','total_energy']].plot.box(title="Train Boxplots")
test_df[['gamma','momentum','total_energy']].plot.box(title="Test Boxplots")
```

```python
train_df.plot.scatter(x='beta', y='gamma', alpha=0.3, title="Gamma vs Beta (Train)")
test_df.plot.scatter(x='beta', y='gamma', alpha=0.3, title="Gamma vs Beta (Test)")
```

```python
train_df.plot.scatter(x='momentum', y='total_energy', alpha=0.3, title="Energy vs Momentum (Train)")
test_df.plot.scatter(x='momentum', y='total_energy', alpha=0.3, title="Energy vs Momentum (Test)")
```

```python
combined = pd.concat([
    train_df[['gamma']].assign(dataset="Train"),
    test_df[['gamma']].assign(dataset="Test")
])

combined.boxplot(by="dataset", column="gamma", grid=False)
```
