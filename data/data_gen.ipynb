```python
N = 10_000
rng = np.random.default_rng(2026)
```

```python
mass = 10 ** rng.uniform(0, 6, N)
force = 10 ** rng.uniform(0, 10, N)
force *= rng.choice([-1.0, 1.0], N)
beta = np.empty(N)
n1 = 7000
n2 = 2000
n3 = 1000
beta[:n1] = rng.uniform(0.0, 0.8, n1)
beta[n1:n1+n2] = rng.uniform(0.8, 0.95, n2)
beta[n1+n2:] = rng.uniform(0.95, 0.999999, n3)
rng.shuffle(beta)
initial_velocity = beta * C
time = 10 ** rng.uniform(-4, 5, N)
outputs = np.empty((N, 8), dtype=np.float64)
```

```python
for i in range(N):

    traj = simulate(
        mass=mass[i],
        force=force[i],
        initial_velocity=initial_velocity[i],
        duration=time[i],
        dt=time[i],
        method="analytic",
    )

    state = traj.final_state()

    outputs[i] = [
        state["position"],
        state["velocity"],
        state["beta"],
        state["gamma"],
        state["momentum"],
        state["kinetic_energy"],
        state["total_energy"],
        state["proper_time"],
    ]
```

```python
data = np.column_stack([
    mass,
    force,
    initial_velocity,
    beta,
    time,
    outputs,
])
np.save(
    "relativistic_extreme_test_10k.npy",
    data,
)
print("Generated:", data.shape)
print("Saved: relativistic_extreme_test_10k.npy")
```

```python
N = 1_000_000

rng = np.random.default_rng(2026)
```

```python
C = 299_792_458.0
mass = 10 ** rng.uniform(0, 6, N)
force = (
    10 ** rng.uniform(0, 10, N)
    * rng.choice([-1.0, 1.0], N)
)
```

```python
beta = np.empty(N)

n1 = int(N * 0.70)
n2 = int(N * 0.20)
n3 = N - n1 - n2

beta[:n1] = rng.uniform(0.0, 0.8, n1)

beta[n1:n1 + n2] = rng.uniform(
    0.8,
    0.95,
    n2
)

beta[n1 + n2:] = rng.uniform(
    0.95,
    0.999999,
    n3
)

rng.shuffle(beta)
```

```python
initial_velocity = beta * C
time = 10 ** rng.uniform(-4, 5, N)
outputs = np.empty(
    (N, 8),
    dtype=np.float64
)
```

```python
for i in range(N):

    traj = simulate(
        mass=mass[i],
        force=force[i],
        initial_velocity=initial_velocity[i],
        duration=time[i],
        dt=time[i],
        method="analytic",
    )

    state = traj.final_state()

    outputs[i] = [
        state["position"],
        state["velocity"],
        state["beta"],
        state["gamma"],
        state["momentum"],
        state["kinetic_energy"],
        state["total_energy"],
        state["proper_time"],
    ]

    if (i + 1) % 10_000 == 0:
        print(f"Generated {i + 1:,} / {N:,}")
```

```python
physics_data = np.column_stack([
    mass,
    force,
    initial_velocity,
    beta,
    time,
    outputs,
])
np.save(
    "physics_data.npy",
    physics_data,
)
```
